**Mean, Median and Mode Calculations**

```python
sleep_map = {
    "Less than 5 hours": 4,
    "5–6 hours": 5.5,
    "6–7 hours": 6.5,
    "7–8 hours": 7.5,
    "More than 8 hours": 9
}

effect_map = {
    "No effect": 1,
    "Slight effect": 2,
    "Moderate effect": 3,
    "Strong effect": 4
}
```

```python
df["Sleep Numeric"] = df["How many hours do you usually sleep per night?\n"].map(sleep_map)
df["Effect Numeric"] = df["How does your sleep schedule affect your academic performance (grades)?"].map(effect_map)

df = df.dropna(subset=["Sleep Numeric", "Effect Numeric"])
```

```python
print("Mean of Sleep Hours =", df["Sleep Numeric"].mean())
print("Median of Sleep Hours =", df["Sleep Numeric"].median())
print("Mode of Sleep Hours =", df["Sleep Numeric"].mode()[0])

print()

print("Mean of Academic Effect =", df["Effect Numeric"].mean())
print("Median of Academic Effect =", df["Effect Numeric"].median())
print("Mode of Academic Effect =", df["Effect Numeric"].mode()[0])
```

Mean,Median,and Mode were calculated only for ordinal and numerical-like variables, pure categorical variables were analyzed using frequencies and visualizations

**Data Cleaning**

```python
df= df.dropna()
df.to_csv("Sleep vs Academic Performance Survey (Responses) - Form Responses 1.csv",index=False)
df= df.dropna(subset=['Sleep Numeric', 'Effect Numeric'])
```

```python
df=df.drop(columns=["Timestamp"])
```

**Data Visualization**

Pie charts


```python

res1 = df['How many hours do you usually sleep per night?\n'].value_counts()


plt.pie(res1, labels=res1.index, autopct='%1.1f%%', colors=['hotpink', 'blue','red','yellow','green'], startangle=90)

plt.title('Sleep Hours')

plt.show()
```

```python
res2=df['Do you believe that your sleep affects your academic performance?'].value_counts()
plt.pie(res2, labels=res2.index, autopct='%1.1f%%', colors=['teal', 'crimson'], startangle=90)

plt.title("Sleep Effect on Academic Performance")
plt.show()
```

```python
res3=df['How does your sleep schedule affect your academic performance (grades)?'].value_counts()
plt.pie(res3, labels=res3.index, autopct='%1.1f%%', colors=['blue', 'yellow','red','hotpink'], startangle=90)

plt.title("The level of sleeping affecting grades")
plt.show()
```

```python
res6=df['If you have not finished studying before an exam, what would you most likely do?'].value_counts()
plt.pie(res6, labels=res6.index, autopct='%1.1f%%', colors=['yellow','pink'], startangle=90)

plt.title("Impact of sleeping on performance in important days")
plt.show()
```

```python
res4=df['How would you describe your overall academic performance?'].value_counts()
plt.pie(res4, labels=res4.index, autopct='%1.1f%%', colors=['blue', 'purple','yellow','pink','red'], startangle=90)

plt.title("Overall Academic Performance")
plt.show()
```

Bar charts

```python
counts = df["How many hours do you usually sleep per night?\n"].value_counts()

categories = counts.index
values = counts.values

plt.bar(categories, values, color=['royalblue', 'darkorange', 'crimson', 'teal', 'seagreen'])

plt.title("Sleep Hours")
plt.xlabel("Hours")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)

plt.show()
```
